
##Zadatak. Naći n-ti broj po redu kojem je zbir cifara jednak 10.


**Naivni pristup:**

▶ Ići redom prirodnim brojevima krenuvši od 1, računati zbir
cifara i bilježiti one sa zbirom koji je jednak 10.


Kompleksnost? Možemo li naprednije?

  
 Ideja. Ispisimo nekoliko prvih brojeva koji zadovoljavaju uslove zadatka:

  Dvocifreni: 19, 28, 37, 46, 55, 64, 73, 82, 91: **19 + 9k (opsti oblik)**

  Trocifreni: 109, 118, 127,...

  Dakle, ispitujemo svaki deveti broj, krenuvsi od 19, i tako dalje...






```python
## NAIVNI pristup

def suma_digits(n):
  sum=0
  while n > 0:
      digit=n%10
      sum+=digit
      n=n//10
  return sum

def naivni_suma_10(n):

  number=1
  count = 0 #pratimo koliko do sada takvih br je nadjeno

  while True:

    if suma_digits(number)==10:
        count+=1

    if count == n:
        return number

    number+=1

##Pokretanje programa:
## Ulaz
n=10000
time_start = tx.time()
trazeni_broj = naivni_suma_10(n)
time_end = tx.time()
##Ispis:
print(f"Broj je nadjen i iznosi {trazeni_broj}; izvrsenje procedure je trajalo {time_end-time_start}")

```

```python
###Inteligentniji pristup:

def sum_10_speedup(n):

   number=19
   count=0
   while True:
       if suma_digits(number)==10:
          count+=1

       if count==n:
           return number
       number+=9



##Pokretanje programa:
## Ulaz
n=10000
time_start = tx.time()
trazeni_broj = sum_10_speedup(n)
time_end = tx.time()
##Ispis:
print(f"Broj je nadjen i iznosi {trazeni_broj}; izvrsenje procedure je trajalo {time_end-time_start}")

#MOZEMO LI DODATNO DA UBRZAMO PROCEDURU?
```
