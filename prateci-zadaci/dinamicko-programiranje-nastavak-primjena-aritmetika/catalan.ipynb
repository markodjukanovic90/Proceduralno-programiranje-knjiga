#Katalanovi brojevi


Pojavljuju se u mnogim interesantnim (kombinatornim)
problemima

▶ Broj različitih puteva sa $2n$ koraka na pravougaonoj mreži od
donjeg lijevog ugla, tj. koordinate $(n - 1, 0)$ do gornjeg desnog
ugla $(0, n - 1)$ koji ne presijeca glavnu dijagonalu (već samo
dodiruje).

▶ Broj permutacija skupa $\{1, \ldots, n\}$ koji ne sadrže patern 123
(ili bilo koji drugi dužine 3): Za $n = 3$, to su permutacije 132,
213, 231, 312 i 321, dakle ima ih 5

▶ Prvih nekoliko Katalanovih brojeva: 1, 1, 2, 5, 14, 42, 132,
429, 1430, 4862,...




##Formula definisanja Katalanovih brojeva

$$C_{n+1} = \sum_{i=0}^n C_i \cdot C_{n-i}, C_0=1.$$

Implementirajmo proceduru za racunanje ovih brojeva:

1) Rekurzivnim nacinom;

2) Koristeci memoizaciju (tj. DP).


Uporediti ova dva pristupa (memorijski).

```python
#Rekurzivni pristup:

def catalan_recursive(n):

  if n==0:
    return 1

  catalan_n=0
  for i in range(n):
    catalan_n += catalan_recursive(i)*catalan_recursive(n-1-i)

  return catalan_n


##poziv (potpuno) rekurzivnog metoda:

n=15

time_start=tx.time()
catalan_n=catalan_recursive(n)
time_end=tx.time()

print(f"{n}-tu katalanov broj je jednak {catalan_n}; rekurzivni pristup zahtjeva vremena: {time_end-time_start}")

```

```python
### DP pristup:  Catalan[i]: cuvamo vrijednost i-tog Katalanovog broja
n=15
Catalan = [-1 for _ in range(n+1)] #inicijalizacija
Catalan[0]=1

def catalan_memoization(n, Catalan):

    if n==0:
      return Catalan[0]

    if Catalan[n] != -1:
       return Catalan[n]

    catalan_n=0
    for i in range(n):
        catalan_n += catalan_memoization(i, Catalan)*catalan_memoization(n-1-i, Catalan)

    Catalan[n] =catalan_n #unisi rezultat u memoriju
    return Catalan[n]

time_start=tx.time()
catalan_n=catalan_memoization(n, Catalan)
time_end=tx.time()

print(f"{n}-tu katalanov broj je jednak {catalan_n}; DP pristup zahtjeva vremena: {time_end-time_start}")
```
