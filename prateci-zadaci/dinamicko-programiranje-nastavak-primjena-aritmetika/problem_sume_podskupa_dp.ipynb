## PROBLEM SUME PODSKUPA 


**Ulaz.** Dat je niz nenegativnih cijelih brojeva i broj *sum*.

**Zadatak.** Da li postoji podskup datih brojeva čija je suma upravo jednaka vrijednosti *sum* ?

Zadatak rijesiti uz pomoc paradigme Dinamickog programiranja.


Primjer. $niz = [3, 34, 4, 12, 5, 2], sum = 9$.
Odgovor. $True$.




*Rjesenje.*     

**Memoizacija.** $DP[i, j]:$ odgovor na pitanje (True/False) da li u prefiksnom nizu $niz[: i]$ postoji podskup cija je suma jednaka $j$.


**Optimalna podstruktura (rekurzija).**  Odluka da li $i$-ti element niza ucestvuje u rjesenju ili ne.

1.   $niz[i-1]$ pripada rjesenju $\Longrightarrow$ isti odgovor kao i $DP[i-1, j-niz[i-1]]$, ako je $j \geq nif[i-1]$
2.   $niz[i-1]$ pripada rjesenju   $\Longrightarrow$ isti odgovor kao i za podproblem (i-1, j), ciji je odgovor u $DP[i-1, j]$

Ako je jedan od odgovora True, onda je odgovor za podproblem jednak $DP[i-1, j]$ ili $DP[i-1, j] \vee DP[i-1, j-niz[i-1]]$, ako je $j \geq niz[i-1]$.

Trivijalni slucajevi: $DP[0, j]=False$, za $sum \geq j\geq 1$, $DP[i, 0]=True$, za $|niz|\geq  i \geq 0 $.

Odgovor se nalazi u $DP[ |niz|,sum]$




```python
## IMPLEMENTACIJA DP pristupa za problem

##instanca
niz=[3, 34, 4, 12, 5, 2]
sum = 9

DP=[] #memoizacija
def init(niz, sum): #INICIJALIZACIJA

    DP = [ [-1]* (sum+1) for _ in range(len(niz) + 1)]
    
    #trivijalni slucajevi:
    for j in range(1,sum+1):
        DP[0][j] = False
    for i in range(len(niz)+1): # DP[0][0]=True
        DP[i][0] = True
    
    return DP  
```

```python
### DP rekurzija (top-down pristup, rekurzivni):

def dynamic_programming(niz, i, j):

    if DP[i][j] != -1:
        return DP[i][j]

    if  j - niz[i-1] >= 0:
        dp_i_j = dynamic_programming(niz, i-1, j-niz[i-1]) or dynamic_programming(niz, i-1, j)
        DP[i][j] = dp_i_j
    else:
        DP[i][j] =  dynamic_programming(niz, i-1, j)

    
    return DP[i][j]
    
```

```python
### POZIV funkcije:
#sum=100
DP = init(niz, sum)

## poziv DP rekurzije:
answer=dynamic_programming(niz, len(niz), sum)

if answer:
  print("Postoji podskup sa datim zbirom")
else:
  print("Ne postoji podskup sa datim zbirom")
```
