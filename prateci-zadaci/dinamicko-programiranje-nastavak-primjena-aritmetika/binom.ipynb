#Binomni koeficijenti

Definicija je data sa:
$$ \binom{n}{k} = \frac{n(n-1) \cdots (n-k+1)}{k!}= \frac{n! \cdot k!}{(n-k)!}$$

Druga, pogodnija rekurzivna definicija je data preko Paskalove jednakosti:

$$\binom{n}{k}=\binom{n-1}{k-1}+\binom{n-1}{k}, \binom{n}{0}=1$$

Izracunati binomni koeficijent $\binom{n}{k}$ koristeci:

1) (Naivni) rekurzivni pristpu

2) DP-pristup



```python
## Rekurzivni pristup:

#instanca
n=25
k=12
#Implementacija (naivni) rekurzivni pristup:
def binom_coef(n, k):
  if n<k:
    return 0

  if k==0:
    return 1

  return binom_coef(n-1, k-1) + binom_coef(n-1, k)

time_start=tx.time()
bin_coef_result = binom_coef(n, k) #poziv rekurzivnog metoda
time_end=tx.time()

print(f"Vrijednost bin. koef. {n} nad {k} je {bin_coef_result}; vrijeme potrebno za izvrsavanje (naivnog) rekurzivnog pristupa je: {time_end-time_start} ")
```

```python
## DP-pristup za racunanje Binomnog koeficijenta:

#instanca
n=25
k=12

#metod inicijalizacije
def init(n, k):
    Binom =[]
    for i in range(n+1):
       row_i=[]
       for j in range(k+1):
           row_i.append(-1)
       Binom.append(row_i)

    return Binom

## Memoizacija:

def binom_memoization(n, k, Binom):

  if Binom[n][k] != -1:
      return Binom[n][k]

  if n<k:
    Binom[n][k] = 0
    return Binom[n][k]

  if k==0:
    Binom[n][k] = 0
    return Binom[n][k]

  binom_n_k = binom_memoization(n-1, k-1, Binom) + binom_memoization(n-1, k, Binom)
  Binom[n][k] = binom_n_k
  return  Binom[n][k]

## Poziv DP pristupa:

Binom = init(n, k)
time_start=tx.time()
bin_coef_result = binom_memoization(n, k, Binom) #poziv rekurzivnog metoda
time_end=tx.time()

print(f"Vrijednost bin. koef. {n} nad {k} je {bin_coef_result}; vrijeme potrebno za izvrsavanje DP pristupa je: {time_end-time_start} ")

```
