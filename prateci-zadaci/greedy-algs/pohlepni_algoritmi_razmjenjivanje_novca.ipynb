**$$\LARGE{\textrm{Pohlepni algoritmi}}$$**


* Konstruktivna grupa algoritma, koja izgradjuje rješenje komponentu po komponentu, proširujući postojeće parcijalno rješenje.
* U svakoj iteraciji, komponenta koja *najviše* obećava gledajuci sve ostale kandidate se bira za proširivanje (parcijalnog) rješenja.
* Dakle, lokalno najbolja odluka se donosi u nadi da se će ta odluka voditi u smjeru globalno najboljeg rješenja. 
* Greške koje se prave lokalizovanim odlukama se ne popravljaju. 
* Lokalne odluke se donose $\textbf{pohlepnom}$ funkcijom, koja zavisi od samog problema (emprijski i na osnovu *zdravog* rasudjivanja)


**Prednosti.**

* Brz pristup (često linearan) u veličini instance; jednostavan za implementaciju 
* Prvi (heuristički) pristup;
* Obično daje rješenja solidnog kvaliteta;

**Slabosti.**

* Rijetko osigurava egzaktnost (optimalnost);
* Osjetljiv na pohlepnu funkciju kojom se bira lokalna odluka




**$$\Large{\textrm{Zadatak 1: Problem razmjenjivanja novca}}$$**


Definicija problema. Na raspolagnju su apoeni od $x_1, x_2,...,x_k$ (KM-ova), a novčanica (kovanica) ovih apoena ima, po pretpostavci, beskonačno. Korisnik ima $X$ (KM-ova) i želi da ih razmijeni.

**Zadatak.** Na koji nacin razmijeniti X (KM-ova) tako da korisnik dobije što manje novčanica (kovanica). 


Skica rješenja. 

Rješenje i komponenta rješenja (definicija):

* rješenje: $k$-niz: $(p_1, p_2, ..., p_k)$, gdje je $p_i \in \mathbb{N}_0$ td. je $\sum_{i=0}^k p_i x_i = X$
* parcijalno rješenje: $(p_{r_1}, p_{r_2}, ... , p_{r_s})$, td. je $\sum_{i=1}^s p_{r_i} x_{r_i} \leq X$ i $\{r_1, ..., r_s\} \in [k]$

**Pohlepna funkcija**:
- prvo razmijeni za makismalno mnogo novcanica maksimalnog apoena -- doprinose smanjenju **ukupnog** **broja** novcanica
Dakle, prvo sortirati apoene po njihovoj veličini. Bez smanjenja opštosti, neka je $x_1 \geq x_2 \geq ... \geq x_k$.

**(Pohlepna) odluka**: neka je $s^p$ parcijalno rješenje, sa ukupnom vrijednošću $X_{p}$. Sljedeća komponenta koja se njemu pridodaje je apoen maksimalne vrijednosti koji do sada nije bio razmatran, recimo da je to $x_{i^*}$. 
* Ako je $x_{i^*} > X- X_{p}$ idemo na naredni apoen maks. vrijednost, a $p_{i^*}=0$;
* Inače, $p_{i^*} =\left  \lfloor \frac{X- X_{p}}{x_{i^*}} \right \rfloor$ kojeg dodajemo u $s^p$, a azuriramo $X^p=X^p + p_{i^*}\cdot x_{p^{i^*}}$
* Ako je $X^P==X$ rješenje je nadjeno ili nastavlajmo dalje dok se svi aopeni ne provjere.




```python
#### Implementacija rješenja ####
```

```python
### instanca 
apoeni=[1, 2, 5, 10, 20, 50, 100]
X=568 

# sort -- opadajuci 
apoeni.sort(reverse=True)
print(apoeni)
```

```python
### razmijeni procedura ###
def razmijeni(X, apoeni):
    #init
    s_p=[] # prazno parcijalno rjesenje
    X_p = 0 #suma parcijalno rjesenje 
    i=0 #i-ti apoen (razmatranje)
    
    while i < len(apoeni):

        if apoeni[i] > X-X_p:
            s_p.append(0)
        else:
            s_p.append((X-X_p)//apoeni[i])
            X_p += ((X-X_p)//apoeni[i])* apoeni[i]

        if X_p == X:
            return True, s_p
        i+=1 #sljedeci apoen 
    
    return False, None

## test:
uspjesno, razmjena_apoeni = razmijeni(X, apoeni)
print(f"Optimalna razmjena se postize na sljedeci nacin {razmjena_apoeni} za koju je potrebno {sum(razmjena_apoeni)} novcanica")

```

Iako pohlepni, ovaj algoritam uvijek osigurava **optimalno** rjesenje.
