**$$\Large{\textrm{Problem rezanja štapa}}$$**

**Problem.** Dat je štap dužine $n$ i lista $p$ sa cijenama štapa dužine $i, 1 \leq i \leq n$.

**Zadatak.** Pronađite optimalan način da isiječete štap na manje štapove da bi se maksimizovao profit. 

Koristiti paradigmu dinamickog programiranja.

**Instanca.** duzina = $[1, 2, 3, 4, 5, 6, 7, 8]$; $p= [1, 5, 8, 9, 10, 17, 17, 20]$. Neka je štap dužine $d=4$.

Rješenje. Rezati štap na dva dijela dužine od po 2. Optimalna zarada je u tom slučaju jednaka $ 5 + 5 = 10$.


**Skica dizajna algoritma.** 

* Memoizacija sa (linearnom) strukturom $DP[i]$: cijena optimalnog rezanja za stap duzine $i$
* Rekurzija: Neka je duzina stapa $k$. Pretpostavimo da smo nasli optimalno rezanje za stap duzine $i$, za sve $1 \leq i<k$. Tada je:
   $$ DP[i]= \max \{ DP[k-i] + p[i-1] \mid i=1, \ldots, k \}.$$
* Trivijalni slucajevi: $DP[0]=0; DP[1]=p[0]$

Rjesenje se nalazi u $DP[d]$.
  


```python
#Implementacija algoritma 

#instanca:
p=[1, 5, 8, 9, 10, 17, 17, 20]
d=4
#init:
DP=[]
def init(p, d):
    
    for i in range(d+1):
        DP.append(-1)
    DP[0]=0
    DP[1]=p[0]

        
```

```python
## Glavna rekurzija (top-down -- memoizacija)

def DP_rezanje_stapa(DP, d):
    
    if DP[d] != -1:
        return DP[d]

    best = 0
    for i in range(1, d+1):
        price_cut_i = DP_rezanje_stapa(d-i) + p[i-1]
        if price_cut_i > best:
            best = price_cut_i
            
    DP[d] = best
    return DP[d]       
        
```

```python
## poziv DP-rekurzije: 
init(DP, d)
best_price_d = DP_rezanje_stapa(DP, d)
print("Najbolja cijena rezanja je: ", best_price_d)
```
