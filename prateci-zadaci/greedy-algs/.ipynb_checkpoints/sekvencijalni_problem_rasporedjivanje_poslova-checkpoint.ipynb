**$$\Large{\textrm{Problem rasporedjivanja poslova}}$$**

**Ulaz.**  Dat je niz poslova, gdje svaki posao ima dedlajn (**krajnje vrijeme** završetka) i odgovarajući **profit** ako je posao završen
prije roka. Za izvršavanje svakog posla je potrebno jedinično vrijeme (recimo, 1h). Vrijeme za izvršavanje svih poslova kreće od istog
trenutka.

**Zadatak.** Na koji nacin maksimizovati ukupni profit ako samo jedan posao može da bude izvršavan u trenutnom (diskretizovanom jediničnom) vremenu. Koristiti paradigmu pohlepnih algoritama. 



**Primjer1 instance problema**

Posao: (deadline, price)

T1:     (4, 20)

T2:     (1, 10)

T3:     (1, 40)

T4:    (1, 30)

**Primjer 2 instance problema**

T1: (2, 100) 

T2: (1, 19)

T3: (2, 27)

T4: (1, 25)

T5: (3, 15)

Rjesenje. {T1, T3, T5}

**Ideja za rjesenje problema.**   Najbolje izvrsiti posao sa najvecom cijenom u **najranijem** slobodnom vremenskom slotu.  

* Parcijalno rjesenje: podskup poslova $s_p$ koji mogu biti izvrseno (nema vremenskog konflikta)

* Prosirenje rjesenja: dodati u $s_p$ posao najvece cijene od preostalih tako da ne narusi vremenski (deadline) niti sekvencijalnost izvrsavanja. Ako takav posao ne postoji, prekinuti pretragu i vratiti rezultat.


```python
## instanca:
JOBS = [(2, 100), (1, 19), (2, 27), (1, 25), (3, 15)]

#sort Jobs acc. to prices
def sort(Jobs):

    jobs_indices=[i for i in range(len(Jobs))]
    jobs_indices.sort(key=lambda x: JOBS[x][1], reverse=True)

    return jobs_indices

def lowest_slot(covered_slots_s_p, deadline_job_r):

    low_slot=-1
    for i in range(deadline_job_r):
        if covered_slots_s_p[i] == False:
            return i 
    
    return low_slot
```

```python
## greedy for the job sequencing problem:
def greedy_job_sequencing(JOBS):

    jobs_sort_indices = sort(JOBS)
    #jobs_sort_indices
    s_p=[]
    covered_slots_s_p=[ False for i in range(max([x[0] for x in JOBS])) ]
    r=0 #jobs 
    while r < len(JOBS):

        deadline_job_r, price_r = JOBS[jobs_sort_indices[r]] 
        #lowest-time available slot
        lowest_available=lowest_slot(covered_slots_s_p, deadline_job_r)
        if lowest_available==-1:
            r +=1
            continue 
        #update partial solution:
        covered_slots_s_p[lowest_available]=True #this slot has covered now
        s_p.append(jobs_sort_indices[r])
        print(s_p)
        r +=1 #next job

    return s_p, covered_slots_s_p
```

```python
### poziv greedy funkcije:
s_p, covered_slots_s_p = greedy_job_sequencing(JOBS)
print(f"Rjesenje je {s_p} za koje je pokriveno {covered_slots_s_p}, dok je zarada { sum([ JOBS[i][1] for i in s_p]) } ")
```
