**$$\Large{\textrm{Problem najdužeg zajedničkog podniza (LCS)}}$$**

**Definicija problema**. String $s$ je podniz stringa $s_1$ akko se $s$ može dobiti brisanjem karaktera iz $s_1$. Npr. $\texttt{acd}$ je podniz stringa $\texttt{abbbcdef}$. 


▶ *Ulaz*. Dva stringa $s_1$ i $s_2$. 

▶ *Izlaz*. String $s$ koji je podniz oba stringa, sa maksimalnom mogućom dužinom.


**Primjer.**

▶ Instanca. $s_1 =\texttt{aatdccddc},s_2 =\texttt{agccddta}$

▶ Rješenje. $s = \texttt{accdd}$. Zašto?

Rijesiti problem uz pomoc $\underline{\textrm{Dinamickog programiranja}}$. 



**Skica za rjesavanje LCS problema.**

*Memoizacija.* Razmotrimo koncept prefiksnog stringa za string $s$. To je neprekidni komad stringa koji se uzet sa pocetka stringa. Npr. ako je $s_1 =\texttt{aatdccddc}$, onda je (jedan) njegov prefiksni string (duzine 4) dat sa $\texttt{aatd}$.
Neka je sa 
* $DP[i][j]$ dato optimalno rjesenje (LCS) prefiksnog stringa stringa $s_1$ duzine $i$ te prefiksnog stringa za string $s_2$.


*Rekurzija.* Pretpostavimo da imamo dva stringa iz primjera odozgo. 

Moze se uociti da vrijedi sljedeca optimalna podrstruktura. Posmatrajmo podproblem definisan prefiksnim stringovima duzine $i=5$ i $j=3$ (podproblem je definisan parom $(5, 3)$) i posmatrajmo manje probleme koji relevantni za konstrukciju njegovog optimalnog rjesenja. To je podproblem $(4, 2)$ jer ako znamo njegovo optimalno rjesenje, optimalno rjesenje  (veceg) podproblema $(5,3)$ moze biti dobijeno njegovom konkatenacijom sa slovom $\texttt{a}$; dakle, duze je za 1. 

Sa druge strane, ako rjesavamo podproblem $(4, 3)$, dva manja pod-problema su relevantna za dobijanje njegovog optimalnog rjesenja (kako krajnja slova nisu ista), a to su $(4, 2)$ i $(3, 3)$ koja se moraju rijesiti -- duze od ova dva rjesenja je i rjesenje veceg pod-problema $(4,3)$. Slicnim rezonovanjem, pretpostavljajuci da se u strukturu $DP[i][j]$ cuva duzina optimalnog rjesenja podproblema $(i, j)$, tada vrijedi
$$DP[i][j]= \begin{cases} DP[i-1][j-1], \textrm{ ako }  s_1[i] =s_2[j], \\ \max \{ DP[i-1][j], DP[i][j-1]\}, s_1[i] \neq s_2[j]  \end{cases}$$
Trivijalni slucajevi su: $DP[i][0]=0=DP[0][j], 0 \leq i \leq |s_1|, 0 \leq j \leq |s_2|$.

Rjesenje se nalazi u rjesavanju podproblema $(|s_1|, |s_2|)$, dakle na $DP[|s_1|][|s_2|]$.



```python
## Rjesenje LCS problema uz pomoc DP-a:
#instanca:
s1='aatdccddc'
s2='agccddta'
#inicijalizacija:
DP =[]
def init(DP, m, n):
    for i in range(m+1):
        D_i=[]
        for j in range(n+1):
            D_i.append(-1)
        DP.append(D_i)
    #trivijalni slucajevi:
    for j in range(n+1):
        DP[0][j]=0
    for i in range(m+1):
        DP[i][0]=0
```

```python
#Glavna rekurzija: iterativno rjesenje (a ne rekurzivno) -- tabuliranje

def LCS_DP(DP, s1, s2):

    for i in range(1, len(s1)+1):
        for j in range(1, len(s2)+1):
            if s1[i-1] == s2[j-1]:  
                DP[i][j]= DP[i-1][j-1]+1
            else:
                DP[i][j] = max(DP[i-1][j], DP[i][j-1])
    
```

```python
## poziv rekurzije 
init(DP, len(s1), len(s2))
LCS_DP(DP, s1, s2)
print("LCS za ", s1, " i ", s2,  " je duzine:",  DP[len(s1)][len(s2)])
```

```python
#Rekonstrukcija rjesenja (backtracking):

def backtrack_LCS(DP, m, n):
    
    solution=[]
    ix=m
    jx=n
    
    while ix>0 and jx>0:

        if s1[ix-1] == s2[jx-1]:
            solution.insert(0, s1[ix])
            ix -=1
            iy -=1
        else:
            if DP[i-1][j] == DP[i][j]:
                ix-=1
            else:
                iy-=1
    return solution 
    

print(f"Rekonstrukcija LCS rjesenja: {backtrack_LCS(DP, len(s1), len(s2))}")


```
