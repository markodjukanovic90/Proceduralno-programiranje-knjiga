**$$\Large{\textrm{Dinamičko programiranje}}$$** 

**Ideja dinamičkog programiranja (DP)**: 

▶ Riješiti problem rekurzivno rasparčavanjem na sitnije probleme
(podprobleme)

▶ Koristiti rješenja podproblema da se konstruiše konačno
rješenje – uočiti **optimalnu podstrukturu**

▶ Svaki podproblem (najviše) jednom rješavan (najčešće su oni
preklapajući)

▶ Rasparčavanje na manje probleme: *rekurzija*

▶ Pamtimo rješenja podproblema: *memoizacija*

**Problem najduzeg puta**. 

Neka je data kvadratna matrica A prirodnih brojeva. Pijun se nalazi u gornjem lijevom
uglu. On se može kretati jedno polje dolje ili jedno polje dijagonalno dolje-desno. Doći do
donjeg reda tako da je suma brojeva na putu **maksimalna**.



$\underline{Primjer}.$ 

**Memoizacija**. $DP(i, j)$: maksimalna udaljenost od pozicije $(i, j)$ do donjeg reda. 

**Optimalna podstruktura** (rekurzija).  $DP(i, j)= \max\{ DP(i-1, j-1), DP(i-1, j)\}+A[i][j]$, ako je $j-1 \geq 0$ inace 
 $DP(i, j)= DP(i-1, j) +A[i][j]$.
 Defaultni slucajevi su $D[0][0]= A[0][0]$. 
 
 Pokrenemo rekurziju da se rijesi svaki podproblem $(n-1, j)$, tj. $DP[n-1][j], j=1, ..., n-1$, tada je
 rjesenje dato sa $\max \{ DP[n-1][j] \mid j=0, ..., n-1 \}$
 


```python
## Rjesenje problema (implementacija)

# instanca problema:

A=[ [5, 2, 1, 10],
    [2, 4, 7, 4],
    [7, 9, 2, 1],
    [3, 2, 2, 1]
  ]
n=len(A[0])


## rekurzija 
DP = [ ] # inicijalizacija: matrica DP sa svim -1

def init(n):
    for i in range(n):
        row = [ ]
        for j in range(n):
            row.append(-1)
        DP.append(row)
    #initializaiton
    DP[0][0] = A[0][0]
    for i in range(n-1):
        for j in range(n):
            if i < j:
                DP[i][j]=-10000
    

def rec(i, j):
    
    if i == 0:
        return DP[0][0]
    #print(i)  
    if DP[i][j] != -1: #Podproblem vec racunat
        return DP[i][j]

    best = rec(i-1, j)
    if j > 0:
        best = max(best, rec(i-1, j-1))
    DP[i][j] = best + A[i][j]
    
    return DP[i][j]
```

```python
### poziv DP rekurzije:
max_path=0

#Racunanje rjesenja (max puta do kraja)
for i in range(n):
    init(n)
    max_put_do_n_m1_i = rec(n-1, i)
    if max_put_do_n_m1_i > max_path:
        max_path = max_put_do_n_m1_i


print(max_path)
```
