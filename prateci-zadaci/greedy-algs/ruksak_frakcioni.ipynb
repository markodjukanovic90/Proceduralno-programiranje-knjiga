**$$\Large{\textrm{Dijeljeni problem ruksaka}}$$**

**Definicija problema**. Neka je dato $n$ proizvoda sa svojim težinama i vrijednostima, predstavljeno vektorima $w = (w_1,...,w_n)$ i
$p = (p_1,..., p_n)$, redom. Neka je $W >0$ kapacitet ruksaka. Proizvodi mogu da se podijele teoretski na bilo koji dio (sol, mlijeko, itd.)

**Zadatak**. Odabrati proizvode i dio tih proizvoda koji se stavljaju u ruksak, tako da se maksimizuje profit proizvoda, pri cemu je kapacitet ruksaka ($W$) zadovoljen (nije narusen). Rijesiti problem pohlepnim pristupom. 



**Rjesenje**. 

Definicija parcijalnog rjesenja i komponente:

* rjesenje: dato $n$-torkom $(l_1, ..., l_n)$ gdje je $l_i \in [0, 1]$ dio proizvoda i koji se ubacuje u ruksak i pri tome je $\sum_{i=1}^n l_i w_i \leq W$
* parcijalno rjesenje:  $(l_{s_1},...,l_{s_m})$, $\{s_1,...,s_m\} \subseteq [n]$, td. $\sum_{i=1}^n l_{s_i} w_{s_i} \leq W$.

Pohlepna funkcija: 
* bazirana na ideji da je bolje uzeti sto vise proizvoda vece cijene, ali manje tezine. Dakle, gledamo odnos $\frac{p_i}{w_i}$, veca vrijednost je bolja.
* dakle, sortiramo proizvode prema gore pomenutom odnosu, bez smanjenja opstosti neka su $1,2,...,n$ upravo redoslijed indeksi proizvoda sotrirani dati kriterijumom.
* 

**Prosirivanje (parcijalnog) rjesenja:**
* $s^p=(l_1, ..., l_r)$ tezine $W_p$; Ako je tezina proizvoda $W_p=W$, nadjedno je optimalno rjesenje;
* Inace, ispitati naredni proizvod sa indekom $r+1$.   Postaviti $l_{r+1}:=\min(\frac{w_i}{W-W_p}, 1)$ te dodati u $s^p$ kao naredni element.



```python
## Implementacija pohlepnog pristupa za dijeljeni problem ruksaka 
```

```python
## instanca:
W=12 
p=[2, 3, 1, 12, 8, 4, 5]
w=[9, 5, 3, 4, 7, 2, 8]
indices=[i for i in range(len(p))]

## sort indices acc. to p_i/w_i

def sort_indices(indices, p, w):

    ratios=[pi/wi for pi, wi in zip(p, w) ]
    indices.sort(key=lambda i: ratios[i], reverse=True) 
    return indices 
```

```python
## pohlepna funkcija za problem:
def greedy_fractional_knapsack(p, w, W):

    s_p=[]
    W_p, r = 0, 0 # starting index (product considered)

    ## sort call:
    indices_sorted=sort_indices(indices, p, w)
    #print(indices_sorted)
    
    while r < len(p):

        if W_p == W:
            return s_p, indices_sorted

        l_r=min( (W-W_p) / w[indices_sorted[r]], 1.0)
        s_p.append(l_r)
        W_p += l_r*w[indices_sorted[r]]
        
        r+=1
    
    return s_p, indices_sorted
    
```

```python
s_p, indices_sorted = greedy_fractional_knapsack(p, w, W)
print(f"Rjesenje je uzeti proizvode {indices_sorted[:len(s_p)]} sa dijelovima: {s_p} ")
print(f"Ukupna cijena rjesenja je  {sum([ p[i]*li for  li, i in zip(s_p, indices_sorted[:len(s_p)])])}")
```

Moze se pokazati da ovakva (pohlepna) strategija vodi do optimalnog rjesenja za dijeljeni problem ruksaka
