**$$\Large{\textrm{Problem pokrivanja skupa}}$$**

**Definicija problema**. Neka je u ulazu data familija podskupova $\mathcal{F}=\{F_1, ...., F_m\}$ skupa (univerzuma) $X$ ($F_i \subseteq X$).

**Zadatak**.  Odabrati minimalan broj skupova iz $\mathcal{F}$ td. njihova unije daje čitav skup $X$ koristeci se strategijom pohlepnog pristupa. 


**Motivacija za problem.** $X$: skup obiljezja (znanje engleskog, matematike, programiranja,...).  Svaki skup u $\mathcal{F}$ predstavlja osobu sa njenim karakteristikama (recimo znanjima iz nekih oblasti). Cilj je izabrati osobe u jednu komisiju tako da je barem jedna osoba u komisija kompetentna za svako obiljezje iz skupa $X$.




**Primjer.** $X = \{1, 2, 3, 4, 5\},$ $\mathcal{F} = \{\{1, 2, 3\}, \{1, 4\}, \{2, 4, 5\}, \{2, 3\}, \{4, 5\}, \{1, 5\}\}$. 
Rješenje je: $S = \{\{1, 2, 3\}, \{2, 4, 5\}\}$. Kardinalnost je $|S| = 2$. 


**Skica rjesenja problema pohlepnom strategijom.**

**Parcijalno rjesenje i rjesenje:**

* Parcijalno $s_p=\{i_1, i_2, ..., i_k \}$, gdje je $F_{i_r}\in \mathcal{F}, r \in \{1, ..., k\}$
* Rjesenje: parcijalno rjesenje $s_p$ za koje je $\bigcup_{r=1}^{|s^p|} F_{i_r}=X$.
* Prosirenje parcijalnog rjesenja:  

**Pohlepna funkcija** (rezonovanje): Sto je veci skup $F_i$ koji ne pokriva vec pokrivene elemente skupa $X$ za dato parcijalno rjesenje $s_p$, prioritet njegovog ubacivanja bi trebalo da je veci. Skor svakog elementa $F_i$ u odnosu na parcijalno rjesenje moze biti dat sa:

$$g(s_p, F_i)= | F_i \cap \cup_{j \in s_p} F_j|.$$

U svakoj iteraciji biramo indeks $i^*$ za koji je $g(s_p, F_i)$ maksimalan pri cemu takav $F_{i^*}$ dodajemo u trenutno (parcijalno) rjesenje dok god unija skupova odredjenih sa $s_p$ nije jednaka $X$.




```python
##### Implementacija pohlepnog pristupa za rjesavanje problema pokrivanja skupa
```

```python
### Instanca:
X=[i for i in range(1, 6) ]
F = [{1, 2, 3}, {1, 4}, {2, 4, 5}, {2, 3}, {4, 5}, {1, 5}]

## greedy g-score in each iteration 
'''
vraca indeks najboljeg skupa F_{best_index} koji ima najveci skup nepokrivenih elemenata 
u odnosu na parcijalno rjesenje s_p
'''
def g_best_score(F, s_p, coverage_s_p):

    best_score, best_index = 0, 0
    
    for i in range(len(F)):
        if i not in s_p:
            g_score_set_i_diff = F[i].difference(coverage_s_p)
            if len(g_score_set_i_diff) > best_score:
                best_score = len(g_score_set_i_diff)
                best_index = i

    return best_score, best_index


def greedy(F, X):

    s_p=[]
    coverage_s_p=set()

    while len(coverage_s_p) != len(X):

        score_best, index_best = g_best_score(F, s_p, coverage_s_p)
        print(score_best, index_best, len(coverage_s_p))
        # update partial sol and coverage:
        s_p.append(index_best)
        coverage_s_p = coverage_s_p.union(F[index_best])
    
    return s_p, coverage_s_p

### poziv greedy

sol, coverage = greedy(F, X)
print(f"Rjesenje je: {sol}  koje pokriva: {coverage}")

```
