```python
import time as tx
```

**$\Large{Fibonacci\ rekurzija}$**


$$f_n = f_{n-1} + f_{n-2},\ f_0 =0, f_1=1$$


Direktna rekurzija (prosli cas): faktor grananja je 2 u svakom pozivu, dakle
KOMPLEKSNOST izvrsavanja  je

$$O(2^n)$$
(dakle, eksponencijalna)

Pitanje:

**MOZEMO LI BOLJE OD TOGA?**


Odgovor: Koristiti metod **MEMOIZACIJE**

Ideja: U FIbb rekurziji, vise puta se rjesava isti (pod)problem (sa istim ulazom). Nakon prvog puta, kada se jedan problem rijesi, staviti njegovo rjesenje u neku strukturu podataka.

U Fib rekurziji, struktura niza je dovoljna za cuvanje podproblema $FIB[i]$: cuva vrijednosti i-tog fibonacijevog niza. Svaki naredni put, ne rjesavati problem rekurzivno, vec ga citati iz odgovarajuce strukture (u ovom slucaj $FIB$ niza)



```python
##### Rekurzija za (obicni) fibonacijev niz:

def fib(n):

  if n==0:
    return 0
  if n==1:
    return 1

  return fib(n-1)+fib(n-2)
```

```python
##### Pokreni rekurziju:
n=35
time_start=tx.time()
fib_number = fib(n)
time_end=tx.time()


print(f"{n}-ti fibonacijev broj je: {fib_number}, potrebno je vrijeme od: {(time_end - time_start)}")
```

```python
#### Fib rekurzija koristeci MEMOIZACIJU (memoization):

def initialization(n, Fib=[]):

    for i in range(n+1):
        Fib.append(-1) #if -1: podproblem do sada nije rijesen rekurzivnim pozivima

    # tivial cases:
    Fib[0] = 0
    Fib[1] = 1

### rekurzija koja ukljucuje memoizaciju:
def fib_memoization(n, Fib):

  if Fib[n] != -1:
    return Fib[n]

  fib_n = fib_memoization(n-1, Fib) + fib_memoization(n-2, Fib)
  Fib[n] = fib_n
  return Fib[n]

```

```python
### poziv funkcije sa memoizacijom:

Fib = []
n=35
initialization(n, Fib)

time_begin=tx.time()
fib_number = fib_memoization(n, Fib)
time_end=tx.time()

print(f"{n}-ti fibonacijev broj je: {fib_number}, potrebno je vrijeme od: {(time_end - time_begin)}")
```
