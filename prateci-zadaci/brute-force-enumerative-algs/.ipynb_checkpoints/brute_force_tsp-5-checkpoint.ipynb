**$$\Large{Enumerativni\ algoritmi}$$**

▶ Naivni, rekurzivni algoritmi, egzaktni algoritmi

▶ Kod teških problema -- eksponencijalno vrijeme izvršavanja  
  -- ponekad, ali vrlo rijetko od pomoći

▶ Primjer takvih algoritama: **brute-force**

**Brute force**

Intuitivna, direktna tehnika za rješavanje problema

▶ bazirana pa definiciji i zahtjevima problema

▶ **Sve moguće kombinacije** problema se enumerišu i ispituju -- među njima najbolja se proglašava za rješenje

▶ Primjer enumeracije: raspoređivanje knjiga na police na sve
načine u cilju optimizovanog popunjavanja prostora u polici


**Prednosti:**


1. **Generički pristup:** nije ograničen na rješavanje specifičnog
problema

2. Idealan za male instance i jednostavnije probleme, za "osjetiti" problem

**Slabosti:**

1. Neefikasan za realne instance

2. Ne narocito "kreativan" algoritam




##Problem  Trgovackog putnika

**Ulaz**: Kompletan graf $G=(V,E)$ sa tezinama $T$ na granama $E$ (tezina grane $ij$ odgovara udaljenosti cvora (grada) $i$ od cvora (grada) $j$).

**Problem**: Trgovac krece iz grada $v$. On zeli da obidje sve gradove (cvorove grafa) i da se vrati na mjesto polaska, tako da obidje sve gradove, a da predjen put (cijena putovanja) bude minimalna.

**Totalna enumeracija**:
Rjesenja (kodiranje): permutacije cvorova npr. $[0\ 2\ 4\ 3\ 1]$ odgovara putanji od mjesta 0 pa do mjesta 2, pa mjesta 4, pa mjesta 3, pa mjesta 1, pa do povratka na mjesto 0. Cijena puta odgovara zbiru udaljenosti od grada 0 do 2, grada 2 do rada 4, ..., i na kraju grada 1 do grada 0.

Neka je $|V|=n$. Bilo koja permutacija elemenata {0, ..., n-1} sa fiksnim prvim cvorom (0) predstavlja moguce rjesenje. Ima ih ukupno $(n-1)!$. Rjesenje sa najmanjom tezinom (zbirom grana) je trazeno rjesenje.

Kodirajmo sada brute-force algoritam za TSP







```python
### Kodiranje enumerativnog algoritma za TSP:
##  0 5 6 9 8
##  5 0 1 2 7
##  6 1 0 4 2
##  9 2 4 0 4
##  8 7 2 4 0

# an instance of TSP
graph = [
    [0, 5, 6, 9, 8],
    [5, 0, 1, 2, 7],
    [6, 1, 0, 4, 2],
    [9, 2, 4, 0, 4],
    [8, 7, 2, 4, 0]
]
```

```python
best_solution_quality=10000000 #initialize
best_solution=[] #initialize

def objective(solution, graph):

    length_path=0
    for i in range(len(solution)-1):
        length_path += graph[solution[i]][solution[i+1]]

    length_path += graph[solution[len(graph)-1]][solution[0]]
    return length_path


def TSP(graph, partial_solution):
    global best_solution_quality
    global best_solution

    if len(partial_solution)==len(graph):#if complete solution

        print(partial_solution)
        solution_length=objective(partial_solution, graph)
        if solution_length < best_solution_quality:
            best_solution_quality = solution_length
            best_solution=partial_solution.copy()
        return

    for i in range(len(graph)):
        if i not in partial_solution:
            partial_solution.append(i)
            TSP(graph, partial_solution)
            partial_solution.remove(i)
```

```python
### call the function:
TSP(graph, [0])
print(f"Best solution is: {best_solution} of the length {best_solution_quality} ")
```
