## Problem (jednodimenzionalnog) ruksaka:

**Ulaz:** Neka je dato $n$ proizvoda sa svojim tezinama $W=(w_1, ..., w_n)$ i sa svojim vrijednostima $P=(p_1, ... , p_n)$. Pretpostavimo da imamo ruksak ciji je kapacitet C.

**Problem:** Koji od proizvoda ubaciti u ruksak tako da cijena ubacenih proizvoda bude najveca ali da kapacitet ruksaka ne bude narusen.



**Enumerativni algoritam (brute force)** za problem ruksaka

Kodiranje rjesenja: binarnim nizom duzine n. Na poziciji i je 1 ako se proizvod i stavlja u ruksak, a 0 inace.

Npr. rjesenje $[1\ 0\ 0\ 1]$ bi znacilo da stavljamo u ruksak prvi i cetvrti proizvod, pod uslovom da je kapacit zadovoljen (tj. da je $w_0+w_3 <=C$).

Enumeracija rjesenja: bilo koji binarni niz duzine n treba da se uzme u obzir, provjeriti ukupnu vrijednost onih koji zadovoljavaju kapacitet ruksaka (mogu da stanu u njega)



```python
### Kodiranje rjesenja Problema ruksaka

## Instanca problema
n=8
W=[10, 2, 12, 4, 7, 3, 9, 1]
P=[2, 1, 4, 7, 4, 5, 8, 6]
C=15
```

```python
### Rekurzija:

best_solution_quality=0
best_solution=[]

## solution quality
def objective(solution, W, P, C):

    weight_all=price_all=0

    for i in range(len(solution)):

        if solution[i]==1:
            weight_all += W[i]
            price_all += P[i]

    if weight_all > C:
        return -1
    else:
        return price_all


### main recursive procedure that enumerates solutions:

def Knapsack(partial_solution, W, P, C):
    global  best_solution_quality
    global best_solution

    if len(partial_solution) == len(W):#complete solution
        print(partial_solution)
        partial_solution_quality = objective(partial_solution, W, P, C )
        if partial_solution_quality > best_solution_quality: # update the best solution
            best_solution_quality = partial_solution_quality
            best_solution = partial_solution.copy()
        #return

    for i in range(2): #include 0 or 1
        #partial_solution.append(i)
        Knapsack(partial_solution + [ i ], W, P, C)
        #partial_solution.pop()
```

```python
## run the code:
Knapsack([], W, P, C)
print(f"The best solution is decoded by {best_solution}, with a quality of {best_solution_quality}")
```
