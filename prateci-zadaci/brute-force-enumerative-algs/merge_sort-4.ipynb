**DIVIDE-and-CONQUER algoritamska paradigma**

Rekurzivna paradigma koja ukljucuje razbijanje problema vise pod-problema, koji se potom rekurzivno rjesavaju.

Sastoji se od nekoliko koraka:
  
1. Dijeljenje problema na vise podproblema X_1, X_2, ..., X_k
2. Rekurzivno rjesavanje svih podproblema (pretp. da smo dobili rjesenja S_1, ..., S_k)
3. Procedure **Combine**(S_1, ... , S_k): kombinovanje rjesenja da bi se dobilo rjesenje pocetnog problema


## Merge sort ##

1. Podijeliti niz na dva (podjednaka) disjunktna dijela
2. Sortirati rekurzivno oba dijela
3. Spojiti oba sortirana dijela u jedan sortirna niz (Combine)

```python
############## Metoda Merge sort

def merge_sort(niz):

    n=len(niz)
    if n <= 1:
      return niz

    niz_left=merge_sort(niz[0:n//2]) # sort rekurzivno lijeva strana
    niz_right=merge_sort(niz[n//2 :]) # sort rekurzivno desna strana

    index_left=0
    index_right=0
    X = []
    while index_left < len(niz_left) and index_right < len(niz_right):

        if niz_left[index_left] < niz_right[index_right]:
            X.append(niz_left[index_left])
            index_left += 1
        else:
            X.append(niz_right[index_right])
            index_right += 1

    if index_left == len(niz_left):
        X += niz_right[index_right :]
    elif index_right == len(niz_right):
        X += niz_left[index_left:]

    return X

```

```python
#### call merge sort procedure:

niz  = [10, 2, 23, 7, 1, -2, 28, 30, 4]

print(merge_sort(niz))

```
