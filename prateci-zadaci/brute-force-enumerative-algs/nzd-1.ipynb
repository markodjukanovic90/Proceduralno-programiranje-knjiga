#### REKURZIVNI ALGORITMI: Euklidov algoritam #######

```python
import time as tx
### Euklidov algoritam za racunanje NZD dva broja NZD(x, y) = NZD(y, x%y) = ...
def NZD(x, y):

  if x < 0 or y < 0:
      raise ValueError("Greska kod unosa brojeva. Provjerite unos!")
      return

  if y == 0:
    return x

  return NZD(y, x % y)


```

```python
### unos dva broja, poziv Euklidovog algoritma:
x = int(input("Unesi prvi broj:"))
y = int(input("Unesi drugi broj:"))

print(f'NZD({x}, {y})= {NZD(x, y)}')
```
