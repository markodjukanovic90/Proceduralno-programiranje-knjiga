**Definicija problema Hanojskih kula:**

Imamo tri štapa i n diskova. Cilj je u što manje poteza
premjestiti diskove sa jednog na drugi štap tako da

▶ Samo se jedan disk pomjera po potezu;

▶ Svaki potez se sastoji od uzimanja jednog diska sa jednog od
štapova i stavljanje na vrh drugog štapa;

▶ Disk ne može da bude smješten na manji disk.



**Strategija rjesavanja**

Strategija:

▶ Uzeti prvih n - 1 diskova sa štapa A, i prebaciti ih na štap B.

▶ Uzeti preostali disk sa štapa A i prebaciti na štap C.

▶ Rekurzivno prebaciti diskove sa štapa B na štap C (koristeći
štap A)


```python
import time as tx
```

```python
############ Problem Hanojskih kula


def hanoi(n, A, B, C): ## sa A slozi na C koristeci B

  if n==1:
    print(f"Prebacujemo disk {n} sa {A} na {C}")
    return
  else:
      hanoi(n-1, A, C, B)
      # prebaci najveci na slobodni
      print(f"Prebacujemo disk {n} sa {A} na {C}")
      hanoi(n-1, B, A, C)
      #print(f"Prebacujemo disk {n} sa {C} na {B}")
```

```python
#### Poziv procedure:
n=int(input("Unesi broj diskova"))
stap1="A"
stap2="B"
stap3="C"

hanoi(n, stap1, stap2, stap3)
```
